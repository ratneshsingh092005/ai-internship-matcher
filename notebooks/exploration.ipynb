# Internship Dataset Exploration
Small, focused EDA for the included internship dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/internships.csv')
df.head()
```

```python
df['title'].value_counts().plot(kind='bar', title='Internships by role')
plt.xticks(rotation=60, ha='right'); plt.tight_layout()
```

```python
df['location'].value_counts().plot(kind='bar', title='Location distribution')
plt.tight_layout()
```

```python
df['remote'].value_counts().rename({True:'Remote', False:'On-site'}).plot(kind='bar', title='Remote vs on-site')
plt.tight_layout()
```

```python
skills = df['skills'].str.split(', ').explode().value_counts().head(15)
skills.plot(kind='bar', title='Common skills'); plt.tight_layout()
```

```python
df['salary_available'] = df['salary'].ne('')
df['salary_available'].value_counts().plot(kind='bar', title='Salary availability'); plt.tight_layout()
```

```python
df['description_words'] = df['description'].str.split().str.len()
df['description_words'].plot(kind='hist', bins=15, title='Description length'); plt.tight_layout()
```
